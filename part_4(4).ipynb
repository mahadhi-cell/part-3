```python
import pandas as pd
df = pd.read_csv("train.csv")
filtered_df = df[
    (df["Sex"] == "female") &
    (df["Age"] >= 18) &
    (df["Age"] <= 35) &
    (df["Pclass"] == 1)
]
filtered_df.to_csv("first_class_women.csv", index=False)
print(filtered_df)

print("\nData has been saved to first_class_women.csv")
```
